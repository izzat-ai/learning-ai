#1) Quyida berilgan datasetni yuklab olib, ustun nomlarini koʻring va ularni maʼno jihatdan tushunarli shaklga keltiring. `Unnamed: 0` ustunini tashlab yuboring!

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"
df = pd.read_csv(url)
df.head()
```

```python
#javobni shu yerda yozing
df = df.drop('Unnamed: 0', axis=1)
df.columns
```

#3) Har bir ustundagi boʻsh qiymatlarni aniqlang va ularni toʻldirish strategiyasini ishlab chiqing.

```python
#javobni shu yerda yozing
df.isnull().sum()
```

```python
# har bir mahsulotni miqdori bo'yicha medianini hisoblab , NaN larni shu qiymat bilan to'ldirish
df['Quantity'] = df["Quantity"].fillna(df.groupby("Item")["Quantity"].transform("median"))
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df.isnull().sum()
```

```python
# mahsulot narxini uning miqdoriga ko'paytirish bilan umumiy narxini aniqlemiz
df['Order Total'] = df['Order Total'].fillna(df['Price']*df['Quantity'])
df.isnull().sum()
```

```python
def get_mode(x):
    m = x.dropna().mode()
    return m.iloc[0] if not m.empty else np.nan

# Kategoriya bo'yicha to'ldirish (transform bilan)
df['Category'] = df['Category'].fillna(df.groupby('Item')['Category'].transform(get_mode))
```

```python
df.isnull().sum()
```

```python
# qolgan NaN larni Price bo'yicha moda topib to'ldirish
price_category_mode = df.groupby('Price')['Category'].transform(get_mode)
df['Category'] = df['Category'].fillna(price_category_mode)
```

```python
df.isnull().sum()
```

```python
# mahsulot nomidagi nanlarni Category va Price kesimida eng ko'p uchragan Item bilan to'ldirish
cat_price_mode = df.groupby(['Category', 'Price'])['Item'].transform(get_mode)
df['Item'] = df['Item'].fillna(cat_price_mode)
```

```python
df['Item'].isnull().sum()
```

```python
# Itemdagi qolgan nanlarni Unknown bilan to'ldirish
df['Item'] = df['Item'].fillna('Unknown')
df['Item'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
# payment methoddagi nanlarni mijoz eng ko'p ishlatgan to'lov usuli bilan to'ldirish
cust_payment_mode = df.groupby('Customer ID')['Payment Method'].transform(get_mode)
df['Payment Method'] = df['Payment Method'].fillna(cust_payment_mode)
df.isnull().sum()
```

#9) Manfiy narx yoki 0 ga teng narx qiymatlarini toping va ularni qanday oʻzgartirish kerakligini hal qiling.

```python
#javobni shu yerda yozing
df[df['Price']<=0]
```

```python
# manfiy va 0 ga teng narxlarni nan qilish
df.loc[df['Price']<=0, 'Price'] = np.nan

# mahsulot nomi bo'yicha guruhlab , median bilan to'ldirish
item_price_mode = df.groupby('Item')['Price'].transform('median')
df['Price'] = df['Price'].fillna(item_price_mode)
df.isnull().sum()
```

```python
df[df['Price']<=0]
```

```python
df['Price'] = df['Price'].fillna(df['Order Total']/df['Quantity'])
df.isnull().sum()
```

```python
df[df['Price']<=0]
```

```python
# manfiy yozilib qolgan narxlarni musbat qilish
df['Price'] = df['Price'].abs()

# 0 yozilib qolgan narxlarni kategoriya bo'yicha narxlarini to'g'irlash
cat_median = df.groupby('Category')['Price'].transform('median')
df['Price'] = np.where(df['Price'] == 0, cat_median, df['Price'])

df['Order Total'] = df['Price']*df['Quantity']

# nan lar sonini tekshirish
df.isnull().sum()
```

#16) Mijozlarning qaysi kunlarda koʻproq xarid qilgani bo‘yicha tahlil qiling.

```python
#javobni shu yerda yozing
df['Day_of_Week'] = df['Order Date'].dt.day_name()
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# 4. Kunlar bo'yicha buyurtmalar soni va jami daromadni hisoblaymiz
daily_analysis = df.groupby('Day_of_Week').agg(
    Buyurtmalar_Soni=('Quantity', 'count'),
    Sotilgan_Mahsulotlar=('Quantity', 'sum'),
    Jami_Daromad=('Order Total', 'sum')
).reindex(days_order)
daily_analysis
```
