**DF ustunlari**

- FTHG (Full Time Home Team Goals): Meybon jamoasining to'liq o'yin (90 daqiqa) davomida urgan gollari soni.   
- FTAG (Full Time Away Team Goals): Mehmon jamoasining to'liq o'yin davomida urgan gollari soni.   
- FTR (Full Time Result): O'yinning yakuniy natijasi:H = Home Win (Meybon jamoa g'alabasi)   A = Away Win (Mehmon jamoa g'alabasi)   D = Draw (Durang)   
- HTHG (Half Time Home Team Goals): Birinchi bo'limda (taymda) mezbon jamoa urgan gollar soni.   
- HTAG (Half Time Away Team Goals): Birinchi bo'limda mehmon jamoa urgan gollar soni.   
- HTR (Half Time Result): Birinchi bo'lim yakuniy natijasi (H, A yoki D)


- HS (Home Shots): Mezbon jamoaning darvoza tomonga yo'llagan barcha zarbalari soni.
- AS (Away Shots): Mehmon jamoaning darvoza tomonga yo'llagan barcha zarbalari soni.
- HST (Home Shots on Target): Mezbon jamoaning aniq (darvoza sohasiga borgan) zarbalari soni.   
- AST (Away Shots on Target): Mehmon jamoaning aniq zarbalari soni.   
- HF (Home Fouls): Mezbon jamoa tomonidan ishlatilgan fol (qoida buzilish)lar soni.   
- AF (Away Fouls): Mehmon jamoa tomonidan ishlatilgan follar soni.   
- HC (Home Corners): Mezbon jamoaning burchak zarbalari (korner) soni.   
- AC (Away Corners): Mehmon jamoaning burchak zarbalari soni

**Guruhlash yordamida NaN larni to'ldirish**

- bu amaliyot uchun ai yordamida oddiyroq DF yaratildi va shu DF bilan ishlandi

```python
data = {
    'Category': [
        'Drinks',
        'Drinks',
        'Drinks',
        'Food',
        'Food',
        'Food',
        'Dessert',
        'Dessert',
    ],
    'Item': [
        'Cola',
        'Cola',
        'Fanta',
        'Burger',
        'Burger',
        'Pizza',
        'Cake',
        'Cake',
    ],
    'Price': [10000, np.nan, 10000, 35000, 35000, np.nan, 20000, np.nan],
    'Portion_Weight': [
        np.nan,
        '0.5L',
        '0.5L',
        '250g',
        '250g',
        '400g',
        '150g',
        np.nan,
    ],
    'Payment_Method': [
        'Card',
        'Card',
        np.nan,
        'Cash',
        'Cash',
        'Card',
        np.nan,
        'Cash',
    ],
}

df = pd.DataFrame(data)
df.head()
```

```python
# narx ustunidagi NaN larni mahsulot nomi bo'yicha guruhlab o'rtacha qiymat bilan to'ldirish
df['Price'] = df['Price'].fillna(df.groupby('Item')['Price'].transform('mean'))
df['Price']
```

```python
df[df.Price.isna()]
```

```python
df.Price = df['Price'].fillna(50000)
```

```python
df['Item'].unique()
```

```python
df.isna().sum()
```

```python
df[df['Portion_Weight'].isna()]
```

```python
df['Portion_Weight'] = df.groupby('Item')['Portion_Weight'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else np.nan))
```

```python
df
```

```python
df['Payment_Method'] = df.groupby('Category')['Payment_Method'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else np.nan))
df
```

**DataFrame ma'lumotlarini to'g'rilash**

- DF ai yordamida yaratiladi

```python
dirty = pd.DataFrame({
    " Ism ": [" Ali", "vali ", "ALI", "Sardor", None],
    "Yosh": ["21", "25", "21", "abc", "30"],
    "Sana": ["2026-01-05", "2026/02/10", "2026-01-05", "2026-03-01", None],
    "Maosh": [500, 1200, 500, 99999, np.nan],
})
dirty
```

- ustun nomlarini tekshirish va to'g'rilash

```python
dirty.columns
```

```python
# ustun nomlaridagi bo'sh joylarni olib tashlash va kichik harflar qilish
dirty.columns = dirty.columns.str.strip().str.lower()
dirty.columns
```

```python
# ism ustunidagi qiymatlarni bir xil korinishga keltirish
dirty['ism'] = dirty['ism'].str.strip().str.title()
dirty['ism']
```

```python
dirty.info()
```

```python
# yosh ustunini numeric qilish va matnli qiymatni NaN qilish
dirty['yosh'] = pd.to_numeric(dirty['yosh'], errors='coerce')
dirty['yosh']
```

```python
dirty['yosh'].dtype
```

```python
# sana ustunini datetime formatiga o'tkizish
dirty['sana'] = pd.to_datetime(dirty['sana'], errors='coerce')
dirty['sana']
```

```python
dirty.isna().sum()
```

```python
# takrorlangan bir xil ma'lumotlarni tekshirish
dirty.duplicated()
```

```python
# takrorlangan bir xil qatorlarni o'chirish
dirty = dirty.drop_duplicates()
dirty
```

```python
# ustun nomlarini o'zgartirish
dirty.rename(columns={'ism':'name', 'yosh':'age', 'sana':'date', 'maosh':'salary'})
```
