- yuqoridagi hulosalardan kelib chiqib yangi ustunlar yaratamiz : `IsSenior` - 40 yoshdan katta bo'lganlarni 1 , kichik bo'lganlarni 0 qilamiz

- `SalaryToBalanceRatio` - balansni maoshga nisbati bo'ladi

```python
# train DF Age ustunini int qilish
df_tr['Age'] = df_tr['Age'].astype(np.int8)
df_tr['Age'].dtype
```

```python
# test DF Age ustunini int qilish
df_te['Age'] = df_te['Age'].astype(np.int8)
df_te['Age'].dtype
```

```python
# IsSenior ustunini yaratish train va test DF lar uchun
df_tr['IsSenior'] = np.where(df_tr['Age']>40, 1,  0)
df_te['IsSenior'] = np.where(df_te['Age']>40, 1, 0)
df_tr.head()
```
