**Ushbu sahifada SimpleImputer - NaN larni turli strategiya yordamida avtomatik to'ldirishni amaliyot qilamiz**

```python
import numpy as np
import pandas as pd
import sklearn
```

```python
df = pd.DataFrame({
    "age": [20, 25, np.nan, 35, 40],
    "salary": [2000, np.nan, 4000, 5000, 100000]
})
df
```

```python
X = df[['age', 'salary']]

# train testlarga bo'lish
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, random_state=42, test_size=0.2)

print("shape train:", X_train.shape)
print("shape test:", X_test.shape)
```

```python
from sklearn.impute import SimpleImputer

# NaN lar bilan ishledigan obyekt yaratish
imputer = SimpleImputer(strategy='median')

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

X_train_imputed
```

```python
# eng ko'p takrorlangan qiymat bilan avtomatik to'ldirish
df_freq = pd.DataFrame({
    "city": ["Tashkent", "Samarkand", "Tashkent", np.nan, "Tashkent"],
    "gender": ["Male", "Female", np.nan, "Male", "Male"]
})
df_freq
```

```python
X2 = df_freq[['city', 'gender']]

X2_train, X2_test = train_test_split(X2, random_state=42, test_size=0.2)

print("shape train:", X2_train.shape)
print("shape test:", X2_test.shape)
```

```python
X2_train
```

```python
imputer2 = SimpleImputer(strategy='most_frequent')

X2_train_imputed = imputer2.fit_transform(X2_train)
X2_test_imputed = imputer2.transform(X2_test)

X2_train_imputed
```

```python
# Unknown bilan avtomatik to'ldirish
df_constant = pd.DataFrame({
    "city": ["Tashkent", "Samarkand", np.nan, "Bukhara", np.nan],
    "job": ["Data Scientist", np.nan, "ML Engineer", "Data Analyst", np.nan]
})
df_constant
```

```python
X3 = df_constant[['city', 'job']]

X3_train, X3_test = train_test_split(X3, random_state=42, test_size=0.2)

print("shape train:", X3_train.shape)
print("shape test:", X3_test.shape)
```

```python
X3_train
```

```python
imputer3 = SimpleImputer(strategy='constant', fill_value='Unknown')

X3_train_imputed = imputer3.fit_transform(X3_train)
X3_test_imputed = imputer3.transform(X3_test)

X3_train_imputed
```
