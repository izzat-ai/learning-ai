###**Feature Scaling**

```python
import numpy as np
import pandas as pd
import sklearn
```

**Standard Scaler bilan ishlash**

```python
df = pd.DataFrame({
    "age": [20, 25, 30, 35, 40],
    "salary": [2000, 3000, 5000, 7000, 10000]
})
df
```

```python
df.describe()
```

- shunchaki amaliyot uchun faqatgina X ni ajratib olamiz , y shart emas

```python
X = df[['age', 'salary']]
type(X)
```

```python
# ma'lumotlarni bo'lish
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, random_state=42, test_size=0.2)

print("shape train:", X_train.shape)
print("shape test:", X_test.shape)
```

```python
# featuresni bir xil masshtabga tushirish
from sklearn.preprocessing import StandardScaler

# obyekt yaratish
scaler = StandardScaler()

# scaling qilish
X_train_scaling = scaler.fit_transform(X_train)
X_test_scaling = scaler.transform(X_test)
```
