**Ushbu sahifada OneHotEncoder - matnli ustunlarni songa o'tkizish amaliyot qilinadi**

```python
import numpy as np
import pandas as pd
import sklearn
```

```python
df = pd.DataFrame({
    "city": [
        "Tashkent",
        "Samarkand",
        "Bukhara",
        "Tashkent",
        "Samarkand",
        "Bukhara"
    ],
    "gender": [
        "Male",
        "Female",
        "Female",
        "Male",
        "Male",
        "Female"
    ]
})
df.shape
```

```python
df
```

```python
# X va y larni ajratish (datasetimiz kichikligi uchun y olinmiydi)
X = df[['city', 'gender']]

from sklearn.model_selection import train_test_split

# train va testlarga bo'lish
X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)

print("shape X_train:", X_train.shape)
print("shape X_test:", X_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder

# obyekt yaratish
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

train_encoded = encoder.fit_transform(X_train)
test_encoded = encoder.transform(X_test)

print("X_train:\n", X_train)
print("\ntrain_encoded:\n", train_encoded)
```

```python
# qaysi kategoriyalarni o'rganganini ko'rish
encoder.categories_
```
