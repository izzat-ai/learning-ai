```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    "Talaba": ["Ali", "Ali", "Vali", "Vali", "Hasan", "Hasan"],
    "Fan": ["Matematika", "Ingliz tili",
            "Matematika", "Ingliz tili",
            "Matematika", "Ingliz tili"],
    "Ball": [90, 85, 78, 92, 88, 80]
})
df
```

```python
# har bir talaba alohida qator bo‘lsin
# fanlar alohida ustunlarga aylansin
# kataklarda talabaning shu fandagi bali tursin
df.pivot(index='Talaba', columns='Fan', values='Ball')
```

```python
sales = pd.DataFrame({
    "Shahar": ["Toshkent", "Toshkent", "Samarqand",
               "Samarqand", "Buxoro", "Buxoro"],
    "Mahsulot": ["Laptop", "Telefon",
                 "Laptop", "Telefon",
                 "Laptop", "Telefon"],
    "Sotuv": [10, 20, 15, 25, 8, 18]
})
sales
```

```python
# shaharlar qatorlarda bo‘lsin;
# mahsulotlar ustunlarda bo‘lsin;
# kataklarda sotuv miqdori ko‘rinsin
sales.pivot(index='Shahar', columns='Mahsulot', values='Sotuv')
```

```python
revenue = pd.DataFrame({
    "Do‘kon": ["A", "A", "A",
               "B", "B", "B",
               "C", "C", "C"],
    "Oy": ["Yanvar", "Fevral", "Mart",
           "Yanvar", "Fevral", "Mart",
           "Yanvar", "Fevral", "Mart"],
    "Daromad": [12000, 15000, 18000,
                10000, 14000, 16000,
                8000, 11000, 13000]
})
revenue
```

```python
# har bir do‘kon alohida qator bo‘lsin;
# oylar alohida ustunlarga aylansin;
# kataklarda tegishli oyning daromadi tursin
revenue.pivot(index="Do‘kon", columns='Oy', values='Daromad')
```

```python
sales = pd.DataFrame({
    "Shahar": [
        "Toshkent", "Toshkent",
        "Toshkent", "Toshkent",
        "Buxoro", "Buxoro",
        "Buxoro", "Buxoro"
    ],
    "Mahsulot": [
        "Laptop", "Laptop",
        "Telefon", "Telefon",
        "Laptop", "Laptop",
        "Telefon", "Telefon"
    ],
    "Sotuv": [5, 7, 10, 12, 3, 5, 8, 10]
})
sales
```

```python
# shaharlar qatorlarda bo‘lsin
# mahsulotlar ustunlarga aylansin
# har bir shaharda har bir mahsulot bo‘yicha o‘rtacha sotuv ko‘rinsin

sales.pivot_table(index='Shahar', columns='Mahsulot', values='Sotuv', aggfunc='mean')
```
