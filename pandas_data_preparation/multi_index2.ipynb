```python
import numpy as np
import pandas as pd
```

**Talabalarning fanlar bo‘yicha natijalari**

```python
data = {
    "Ball": [86, 91, 78, 88, 95, 82, 76, 89],
    "Davomat": [92, 96, 85, 90, 98, 88, 80, 94]
}

index = pd.MultiIndex.from_tuples(
    [
        ("Ali", "Matematika"),
        ("Ali", "Python"),
        ("Vali", "Matematika"),
        ("Vali", "Python"),
        ("Hasan", "Matematika"),
        ("Hasan", "Python"),
        ("Sardor", "Matematika"),
        ("Sardor", "Python")
    ],
    names=["Talaba", "Fan"]
)

df = pd.DataFrame(data, index=index)
df
```

```python
# indeksni tekshirish
df.index
```

```python
# Faqat Alining barcha fanlar bo‘yicha ma’lumotlarini olish
df.loc['Ali']
```

```python
# Faqat Python fanidan barcha talabalar natijasini olish
df.xs('Python', level=1)
```

```python
# Alining faqat Python fanidagi Ball qiymatini olish
df.loc['Ali', 'Python']['Ball']
```

```python
# Matematika fanidan barcha talabalarni va ularning Ball hamda Davomat qiymatlarini olish
df.xs('Matematika', level=1)
```

**Do‘kon savdolari ma'lumotlari**

```python
data = {
    "Sotuv": [120, 150, 90, 200, 180, 110, 250, 210],
    "Daromad": [1200, 1500, 900, 2000, 1800, 1100, 2500, 2100]
}

index = pd.MultiIndex.from_tuples(
    [
        ("Toshkent", "Yanvar"),
        ("Toshkent", "Fevral"),
        ("Toshkent", "Mart"),
        ("Samarqand", "Yanvar"),
        ("Samarqand", "Fevral"),
        ("Samarqand", "Mart"),
        ("Buxoro", "Yanvar"),
        ("Buxoro", "Fevral")
    ],
    names=["Shahar", "Oy"]
)

sales = pd.DataFrame(data, index=index)
sales
```

```python
# Faqat Toshkent shahrining barcha oylaridagi ma’lumotlarni olish
sales.loc['Toshkent']
```

```python
# Faqat Fevral oyidagi barcha shaharlar ma’lumotlarini olish
sales.xs('Fevral', level=1)
```

```python
# Samarqand Mart ma'lumotlarini oling
sales.loc['Samarqand', 'Mart']
```

```python
# Buxoro shahrining barcha ma’lumotlarini oling
sales.loc['Buxoro']
```
