```python
import numpy as np
import pandas as pd
```

**Talabalarning fanlar bo‘yicha natijalari**

```python
data = {
    "Ball": [86, 91, 78, 88, 95, 82, 76, 89],
    "Davomat": [92, 96, 85, 90, 98, 88, 80, 94]
}

index = pd.MultiIndex.from_tuples(
    [
        ("Ali", "Matematika"),
        ("Ali", "Python"),
        ("Vali", "Matematika"),
        ("Vali", "Python"),
        ("Hasan", "Matematika"),
        ("Hasan", "Python"),
        ("Sardor", "Matematika"),
        ("Sardor", "Python")
    ],
    names=["Talaba", "Fan"]
)

df = pd.DataFrame(data, index=index)
df
```

```python
# indeksni tekshirish
df.index
```

```python
# Faqat Alining barcha fanlar bo‘yicha ma’lumotlarini olish
df.loc['Ali']
```

```python
# Faqat Python fanidan barcha talabalar natijasini olish
df.xs('Python', level=1)
```

```python
# Alining faqat Python fanidagi Ball qiymatini olish
df.loc['Ali', 'Python']['Ball']
```

```python
# Matematika fanidan barcha talabalarni va ularning Ball hamda Davomat qiymatlarini olish
df.xs('Matematika', level=1)
```

**Do‘kon savdolari ma'lumotlari**

```python
data = {
    "Sotuv": [120, 150, 90, 200, 180, 110, 250, 210],
    "Daromad": [1200, 1500, 900, 2000, 1800, 1100, 2500, 2100]
}

index = pd.MultiIndex.from_tuples(
    [
        ("Toshkent", "Yanvar"),
        ("Toshkent", "Fevral"),
        ("Toshkent", "Mart"),
        ("Samarqand", "Yanvar"),
        ("Samarqand", "Fevral"),
        ("Samarqand", "Mart"),
        ("Buxoro", "Yanvar"),
        ("Buxoro", "Fevral")
    ],
    names=["Shahar", "Oy"]
)

sales = pd.DataFrame(data, index=index)
sales
```

```python
# Faqat Toshkent shahrining barcha oylaridagi ma’lumotlarni olish
sales.loc['Toshkent']
```

```python
# Faqat Fevral oyidagi barcha shaharlar ma’lumotlarini olish
sales.xs('Fevral', level=1)
```

```python
# Samarqand Mart ma'lumotlarini oling
sales.loc['Samarqand', 'Mart']
```

```python
# Faqat Daromad ustunidan foydalanib, barcha shaharlar bo‘yicha ma'lumot olish
sales['Daromad']
```

```python
# Buxoro shahrining barcha ma’lumotlarini oling
sales.loc['Buxoro']
```

**`stack`/`unstack` amaliyoti**

```python
data2 = {
    "Q1": [120, 150, 100, 130],
    "Q2": [140, 170, 110, 145],
    "Q3": [160, 180, 125, 155],
    "Q4": [190, 210, 140, 170]
}

index2 = pd.MultiIndex.from_tuples(
    [
        ("Ali", "Laptop"),
        ("Ali", "Telefon"),
        ("Vali", "Laptop"),
        ("Vali", "Telefon")
    ],
    names=["Sotuvchi", "Mahsulot"]
)

df2 = pd.DataFrame(data2, index=index2)

df2
```

```python
# Jadvalni shunday ko‘rinishga keltiringki, choraklar ham indeks darajasiga o'tsin
df2_stack = df2.stack()
df2_stack
```

```python
# Hosil bo‘lgan jadvaldan foydalanib, uni yana dastlabki ko‘rinishga qaytaring
df2_stack.unstack()
```

```python
# Faqat Laptop mahsulotlari bo‘yicha choraklar kesimidagi ma’lumotlarni ko‘rishga harakat qiling
df2.xs('Laptop', level=1)
```

```python
# Jadvalni shunday o‘zgartiringki, Mahsulot indeks emas, ustunlar darajasida joylashsin
df2.unstack(level=1)
```

**4-topshiriq**

```python
data3 = {
    "Marketing": [12000, 15000, 18000, 14000, 17000, 21000, 13000, 16000],
    "IT": [25000, 28000, 30000, 27000, 31000, 35000, 24000, 29000],
    "HR": [8000, 9000, 9500, 8500, 10000, 11000, 7800, 9200]
}

index3 = pd.MultiIndex.from_tuples(
    [
        ("Toshkent", 2023),
        ("Toshkent", 2024),
        ("Samarqand", 2023),
        ("Samarqand", 2024),
        ("Buxoro", 2023),
        ("Buxoro", 2024),
        ("Fargona", 2023),
        ("Fargona", 2024)
    ],
    names=["Hudud", "Yil"]
)

expenses = pd.DataFrame(data3, index=index3)
expenses
```

```python
# Faqat 2024-yil ma’lumotlarini ajratib oling
expenses.xs(2024, level=1)
```

```python
# Faqat Toshkent va Buxoro hududlarini ajratib oling
expenses.loc[['Toshkent', 'Buxoro']]
```

```python
# Samarqand — 2023 yil ma’lumotlarini oling
expenses.loc['Samarqand', 2023]
```

```python
# Yil alohida indeks darajasi sifatida emas, ustunlar tomonida joylashsin
expenses.unstack(level=1)
```

**5-topshiriq**

- Quyidagi DataFrame kompaniyaning hudud → yil → mahsulot kesimidagi savdo ma’lumotlari

```python
data4 = {
    "Sotuv": [
        120, 150, 200,
        180, 210, 250,
        100, 130, 170,
        160, 190, 230
    ],
    "Buyurtmalar": [
        12, 15, 20,
        18, 21, 25,
        10, 13, 17,
        16, 19, 23
    ]
}

index4 = pd.MultiIndex.from_tuples(
    [
        ("Toshkent", 2023, "Laptop"),
        ("Toshkent", 2023, "Telefon"),
        ("Toshkent", 2023, "Tablet"),

        ("Toshkent", 2024, "Laptop"),
        ("Toshkent", 2024, "Telefon"),
        ("Toshkent", 2024, "Tablet"),

        ("Samarqand", 2023, "Laptop"),
        ("Samarqand", 2023, "Telefon"),
        ("Samarqand", 2023, "Tablet"),

        ("Samarqand", 2024, "Laptop"),
        ("Samarqand", 2024, "Telefon"),
        ("Samarqand", 2024, "Tablet")
    ],
    names=["Hudud", "Yil", "Mahsulot"]
)

sales = pd.DataFrame(data4, index=index4)
sales
```

```python
# Faqat Toshkent → 2024 → Telefon ma’lumotlarini oling
sales.loc[('Toshkent', 2024, 'Telefon')]
```

```python
# Faqat Samarqand hududidagi barcha ma’lumotlarni oling
sales.loc['Samarqand']
```

```python
# Faqat 2023-yil ma’lumotlarini ajrating
sales.xs(2023, level=1)
```

```python
# Mahsulot indeksdan ustunlar tomoniga o'tadigan ko‘rinish hosil qiling
sales_unstack = sales.unstack(level=2)
sales_unstack
```

```python
# Endi Yil ham alohida ustunlar darajasida ko‘rinadigan qilib jadvalni o‘zgartiring
sales_unstack.unstack(level=1)
```

```python
sales_unstack.stack(future_stack=True)
```
