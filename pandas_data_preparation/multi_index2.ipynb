**Talabalarning fanlar bo‘yicha natijalari**

```python
data = {
    "Ball": [86, 91, 78, 88, 95, 82, 76, 89],
    "Davomat": [92, 96, 85, 90, 98, 88, 80, 94]
}

index = pd.MultiIndex.from_tuples(
    [
        ("Ali", "Matematika"),
        ("Ali", "Python"),
        ("Vali", "Matematika"),
        ("Vali", "Python"),
        ("Hasan", "Matematika"),
        ("Hasan", "Python"),
        ("Sardor", "Matematika"),
        ("Sardor", "Python")
    ],
    names=["Talaba", "Fan"]
)

df = pd.DataFrame(data, index=index)
df
```

**`stack`/`unstack` amaliyoti**

```python
data2 = {
    "Q1": [120, 150, 100, 130],
    "Q2": [140, 170, 110, 145],
    "Q3": [160, 180, 125, 155],
    "Q4": [190, 210, 140, 170]
}

index2 = pd.MultiIndex.from_tuples(
    [
        ("Ali", "Laptop"),
        ("Ali", "Telefon"),
        ("Vali", "Laptop"),
        ("Vali", "Telefon")
    ],
    names=["Sotuvchi", "Mahsulot"]
)

df2 = pd.DataFrame(data2, index=index2)

df2
```

```python
# Jadvalni shunday ko‘rinishga keltiringki, choraklar ham indeks darajasiga o'tsin
df2_stack = df2.stack()
df2_stack
```

```python
# Hosil bo‘lgan jadvaldan foydalanib, uni yana dastlabki ko‘rinishga qaytaring
df2_stack.unstack()
```

```python
# Faqat Laptop mahsulotlari bo‘yicha choraklar kesimidagi ma’lumotlarni ko‘rishga harakat qiling
df2.xs('Laptop', level=1)
```

```python
# Jadvalni shunday o‘zgartiringki, Mahsulot indeks emas, ustunlar darajasida joylashsin
df2.unstack(level=1)
```

**4-topshiriq**

```python
data3 = {
    "Marketing": [12000, 15000, 18000, 14000, 17000, 21000, 13000, 16000],
    "IT": [25000, 28000, 30000, 27000, 31000, 35000, 24000, 29000],
    "HR": [8000, 9000, 9500, 8500, 10000, 11000, 7800, 9200]
}

index3 = pd.MultiIndex.from_tuples(
    [
        ("Toshkent", 2023),
        ("Toshkent", 2024),
        ("Samarqand", 2023),
        ("Samarqand", 2024),
        ("Buxoro", 2023),
        ("Buxoro", 2024),
        ("Fargona", 2023),
        ("Fargona", 2024)
    ],
    names=["Hudud", "Yil"]
)

expenses = pd.DataFrame(data3, index=index3)
expenses
```

**5-topshiriq**

- Quyidagi DataFrame kompaniyaning hudud → yil → mahsulot kesimidagi savdo ma’lumotlari

```python
data4 = {
    "Sotuv": [
        120, 150, 200,
        180, 210, 250,
        100, 130, 170,
        160, 190, 230
    ],
    "Buyurtmalar": [
        12, 15, 20,
        18, 21, 25,
        10, 13, 17,
        16, 19, 23
    ]
}

index4 = pd.MultiIndex.from_tuples(
    [
        ("Toshkent", 2023, "Laptop"),
        ("Toshkent", 2023, "Telefon"),
        ("Toshkent", 2023, "Tablet"),

        ("Toshkent", 2024, "Laptop"),
        ("Toshkent", 2024, "Telefon"),
        ("Toshkent", 2024, "Tablet"),

        ("Samarqand", 2023, "Laptop"),
        ("Samarqand", 2023, "Telefon"),
        ("Samarqand", 2023, "Tablet"),

        ("Samarqand", 2024, "Laptop"),
        ("Samarqand", 2024, "Telefon"),
        ("Samarqand", 2024, "Tablet")
    ],
    names=["Hudud", "Yil", "Mahsulot"]
)

sales = pd.DataFrame(data4, index=index4)
sales
```

```python
# Faqat Toshkent → 2024 → Telefon ma’lumotlarini oling
sales.loc[('Toshkent', 2024, 'Telefon')]
```

```python
# Faqat Samarqand hududidagi barcha ma’lumotlarni oling
sales.loc['Samarqand']
```

```python
# Faqat 2023-yil ma’lumotlarini ajrating
sales.xs(2023, level=1)
```

```python
# Mahsulot indeksdan ustunlar tomoniga o'tadigan ko‘rinish hosil qiling
sales_unstack = sales.unstack(level=2)
sales_unstack
```

```python
sales_unstack.index.names
```

```python
# Toshkentda 2023 va 2024-yillarda Laptop, Telefon va Tablet savdolarini bir qarashda taqqoslash
sales_unstack.loc['Toshkent']
```

```python
sales_unstack.stack(future_stack=True)
```

**pivot_table**

- Talabalar natijasi ma'lumotlari

```python
data = {
    "Talaba": [
        "Ali", "Ali", "Ali",
        "Vali", "Vali", "Vali",
        "Hasan", "Hasan", "Hasan",
        "Sardor", "Sardor", "Sardor"
    ],
    "Fan": [
        "Matematika", "Python", "Statistika",
        "Matematika", "Python", "Statistika",
        "Matematika", "Python", "Statistika",
        "Matematika", "Python", "Statistika"
    ],
    "Ball": [
        85, 92, 78,
        76, 88, 81,
        91, 95, 89,
        80, 90, 84
    ],
    "Davomat": [
        90, 95, 85,
        80, 88, 90,
        96, 98, 94,
        85, 92, 89
    ]
}

df = pd.DataFrame(data)
df
```

```python
# Har bir talabaning har bir fan bo‘yicha Ball qiymatini alohida ko‘rsatadigan jadval hosil qiling
df_pivot = pd.pivot_table(df, index='Talaba', columns='Fan', values='Ball')
df_pivot
```

```python
# Har bir talabaning o‘rtacha Ballini hisoblang
df_pivot.mean(axis=1)
```

```python
# Har bir fanning o‘rtacha Ballini hisoblang
df_pivot.mean(axis=0)
```

**Savdo tahlili ma'lumotlari**

```python
data = {
    "Shahar": [
        "Toshkent", "Toshkent", "Toshkent",
        "Toshkent", "Toshkent", "Toshkent",
        "Samarqand", "Samarqand", "Samarqand",
        "Samarqand", "Samarqand", "Samarqand"
    ],
    "Oy": [
        "Yanvar", "Yanvar", "Fevral",
        "Fevral", "Mart", "Mart",
        "Yanvar", "Yanvar", "Fevral",
        "Fevral", "Mart", "Mart"
    ],
    "Mahsulot": [
        "Laptop", "Telefon", "Laptop",
        "Telefon", "Laptop", "Telefon",
        "Laptop", "Telefon", "Laptop",
        "Telefon", "Laptop", "Telefon"
    ],
    "Sotuv": [
        120, 180, 150,
        200, 170, 220,
        90, 140, 110,
        160, 130, 180
    ],
    "Daromad": [
        12000, 18000, 15000,
        20000, 17000, 22000,
        9000, 14000, 11000,
        16000, 13000, 18000
    ]
}

sales = pd.DataFrame(data)
sales
```

```python
# Har bir Shahar × Mahsulot bo‘yicha jami Sotuvni ko‘rsatadigan jadval hosil qiling
sales.pivot_table(index='Mahsulot', columns = 'Oy', values='Sotuv', aggfunc='sum')
```

```python
# Har bir Shahar × Oy bo‘yicha jami Daromadni ko‘rsating
sales.pivot_table(index='Shahar', columns='Oy', values='Daromad', aggfunc='sum')
```

```python
# Har bir Shahar × Mahsulot bo‘yicha o‘rtacha Sotuvni hisoblang
sales.pivot_table(index='Shahar', columns='Mahsulot', values='Sotuv', aggfunc='mean')
```

```python
toshkent = sales[sales['Shahar'] == 'Toshkent']
toshkent.pivot_table(index='Oy', columns='Shahar', values='Sotuv', aggfunc='sum')
```

```python
# Jadvalni shunday tuzingki, Shahar va Mahsulot birgalikda indeksning ikki darajasini tashkil qilsin, ustunlarda esa Oy bo‘lsin
pd.pivot_table(sales, columns='Oy', index=['Shahar', 'Mahsulot'])
```
