```python
import numpy as np
import pandas as pd
```

**Do'kon savdolari ma'lumotlari**

```python
columns = pd.MultiIndex.from_tuples(
    [
        ("Sotuv", "Laptop"),
        ("Sotuv", "Telefon"),
        ("Sotuv", "Tablet"),
        ("Daromad", "Laptop"),
        ("Daromad", "Telefon"),
        ("Daromad", "Tablet")
    ],
    names=["Ko‘rsatkich", "Mahsulot"]
)

data = [
    [120, 180, 90, 12000, 18000, 9000],
    [150, 210, 110, 15000, 21000, 11000],
    [170, 230, 130, 17000, 23000, 13000],
    [140, 190, 100, 14000, 19000, 10000]
]

index = ["Toshkent", "Samarqand", "Buxoro", "Fargona"]

sales = pd.DataFrame(data, index=index, columns=columns)
sales
```

```python
# Faqat Sotuv ko‘rsatkichiga tegishli barcha mahsulotlarni oling
sales['Sotuv']
```

```python
# Faqat Laptop mahsulotining Sotuv va Daromad ma’lumotlarini oling
sales.xs('Laptop', level='Mahsulot', axis=1)
```

```python
# Faqat Telefon uchun Daromad qiymatlarini oling
sales[('Daromad', 'Telefon')]
```

```python
# Toshkent va Buxoro hududlarining faqat Sotuv ma’lumotlarini oling
sales.loc[['Toshkent', 'Buxoro'], 'Sotuv']
```

**2-topshiriq**

```python
columns = pd.MultiIndex.from_tuples(
    [
        (2023, "Sotuv"),
        (2023, "Daromad"),
        (2023, "Buyurtma"),
        (2024, "Sotuv"),
        (2024, "Daromad"),
        (2024, "Buyurtma")
    ],
    names=["Yil", "Ko‘rsatkich"]
)

data = [
    [120, 12000, 12, 150, 15000, 15],
    [100, 10000, 10, 140, 14000, 14],
    [90, 9000, 9, 130, 13000, 13],
    [110, 11000, 11, 160, 16000, 16]
]

df = pd.DataFrame(
    data,
    index=["Toshkent", "Samarqand", "Buxoro", "Fargona"],
    columns=columns
)
df
```

```python
# Faqat Daromad ko‘rsatkichining 2023 va 2024-yillardagi qiymatlarini oling
df.xs('Daromad', level='Ko‘rsatkich', axis=1)
```

```python
# Faqat 2024-yil Sotuv ma’lumotlarini oling
df[(2024, 'Sotuv')]
```

```python
# Toshkent uchun 2023 va 2024-yillardagi barcha ma’lumotlarni oling
df.loc['Toshkent']
```

```python
# Buxoro uchun faqat 2024-yil ma’lumotlarini oling
df.loc['Buxoro', 2024]
```

**3-topshiriq**

```python
columns = pd.MultiIndex.from_tuples(
    [
        ("Laptop", "Sotuv"),
        ("Laptop", "Daromad"),
        ("Laptop", "Buyurtma"),
        ("Telefon", "Sotuv"),
        ("Telefon", "Daromad"),
        ("Telefon", "Buyurtma"),
        ("Tablet", "Sotuv"),
        ("Tablet", "Daromad"),
        ("Tablet", "Buyurtma")
    ],
    names=["Mahsulot", "Ko‘rsatkich"]
)

data = [
    [120, 12000, 12, 180, 18000, 18, 90, 9000, 9],
    [150, 15000, 15, 210, 21000, 21, 110, 11000, 11],
    [100, 10000, 10, 160, 16000, 16, 80, 8000, 8],
    [140, 14000, 14, 200, 20000, 20, 100, 10000, 10]
]

df3 = pd.DataFrame(
    data,
    index=["Toshkent", "Samarqand", "Buxoro", "Fargona"],
    columns=columns
)
df3
```

```python
# Ustunlarning darajalarini shunday almashtiringki, Ko‘rsatkich birinchi daraja, Mahsulot ikkinchi daraja bo‘lsin
df3_swap = df3.swaplevel("Mahsulot", "Ko‘rsatkich", axis=1)
df3_swap
```

```python
# Hosil bo‘lgan jadvalni ustunlar bo‘yicha tartibli holatga keltiring
df3_swap.sort_index(axis=1, level=0)
```

**4-topshiriq**

```python
columns = pd.MultiIndex.from_tuples(
    [
        ("2024", "Laptop", "Sotuv"),
        ("2024", "Laptop", "Daromad"),
        ("2024", "Telefon", "Sotuv"),
        ("2024", "Telefon", "Daromad"),

        ("2023", "Laptop", "Sotuv"),
        ("2023", "Laptop", "Daromad"),
        ("2023", "Telefon", "Sotuv"),
        ("2023", "Telefon", "Daromad")
    ],
    names=["Yil", "Mahsulot", "Ko‘rsatkich"]
)

data = [
    [150, 15000, 220, 22000, 120, 12000, 180, 18000],
    [170, 17000, 250, 25000, 140, 14000, 200, 20000],
    [130, 13000, 190, 19000, 100, 10000, 160, 16000],
    [160, 16000, 230, 23000, 130, 13000, 190, 19000]
]

df4 = pd.DataFrame(
    data,
    index=["Toshkent", "Samarqand", "Buxoro", "Fargona"],
    columns=columns
)
df4
```

```python
# Faqat 2024-yil ma’lumotlarini oling
df4['2024']
```

```python
# Faqat 2023-yildagi Telefon ma’lumotlarini oling
df4.xs('Telefon', level='Mahsulot', axis=1)['2023']
```

```python
# Faqat 2024-yildagi Laptop → Sotuv qiymatlarini oling
df4[('2024', 'Laptop', 'Sotuv')]
```

```python
# Ustunlar tartibini shunday o‘zgartiringki, Ko‘rsatkich → Mahsulot → Yil tartibida bo‘lsin
df4_swap = df4.swaplevel(2, 0, axis=1).sort_index(axis=1)
df4_swap
```

```python
# Yakuniy jadvaldan faqat Sotuv ma’lumotlarini ajrating
df4_swap['Sotuv']
```

```python
# Yakuniy jadvalda Laptop va Telefon sotuvlarini 2023 va 2024 yillar bo‘yicha bir qarashda taqqoslash mumkin bo‘lsin
sotuv = df4_swap['Sotuv']
sotuv.stack(level='Mahsulot', future_stack=True)
```
