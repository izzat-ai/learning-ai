```python
import numpy as np
import pandas as pd
```

**Do'kon savdolari ma'lumotlari**

```python
columns = pd.MultiIndex.from_tuples(
    [
        ("Sotuv", "Laptop"),
        ("Sotuv", "Telefon"),
        ("Sotuv", "Tablet"),
        ("Daromad", "Laptop"),
        ("Daromad", "Telefon"),
        ("Daromad", "Tablet")
    ],
    names=["Ko‘rsatkich", "Mahsulot"]
)

data = [
    [120, 180, 90, 12000, 18000, 9000],
    [150, 210, 110, 15000, 21000, 11000],
    [170, 230, 130, 17000, 23000, 13000],
    [140, 190, 100, 14000, 19000, 10000]
]

index = ["Toshkent", "Samarqand", "Buxoro", "Fargona"]

sales = pd.DataFrame(data, index=index, columns=columns)
sales
```

```python
# Faqat Sotuv ko‘rsatkichiga tegishli barcha mahsulotlarni oling
sales['Sotuv']
```

```python
# Faqat Laptop mahsulotining Sotuv va Daromad ma’lumotlarini oling
sales.xs('Laptop', level='Mahsulot', axis=1)
```

```python
# Faqat Telefon uchun Daromad qiymatlarini oling
sales[('Daromad', 'Telefon')]
```

```python
# Toshkent va Buxoro hududlarining faqat Sotuv ma’lumotlarini oling
sales.loc[['Toshkent', 'Buxoro'], 'Sotuv']
```

**2-topshiriq**

```python
columns = pd.MultiIndex.from_tuples(
    [
        (2023, "Sotuv"),
        (2023, "Daromad"),
        (2023, "Buyurtma"),
        (2024, "Sotuv"),
        (2024, "Daromad"),
        (2024, "Buyurtma")
    ],
    names=["Yil", "Ko‘rsatkich"]
)

data = [
    [120, 12000, 12, 150, 15000, 15],
    [100, 10000, 10, 140, 14000, 14],
    [90, 9000, 9, 130, 13000, 13],
    [110, 11000, 11, 160, 16000, 16]
]

df = pd.DataFrame(
    data,
    index=["Toshkent", "Samarqand", "Buxoro", "Fargona"],
    columns=columns
)
df
```

```python
# Faqat 2023-yil ma’lumotlarini oling
df[2023]
```

```python
# Faqat Daromad ko‘rsatkichining 2023 va 2024-yillardagi qiymatlarini oling
df.xs('Daromad', level='Ko‘rsatkich', axis=1)
```

```python
# Faqat 2024-yil Sotuv ma’lumotlarini oling
df[2024]
```

```python
# Toshkent uchun 2023 va 2024-yillardagi barcha ma’lumotlarni oling
df.loc['Toshkent']
```

```python
# Buxoro uchun faqat 2024-yil ma’lumotlarini oling
df.loc['Buxoro', 2024]
```
