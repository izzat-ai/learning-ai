```python
import numpy as np
import pandas as pd
```

```python
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],
    "name": ["Ali", "Vali", "Hasan", "Sardor", "Madina"],
    "city": ["Toshkent", "Samarqand", "Buxoro", "Toshkent", "Andijon"]
})

orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6],
    "customer_id": [101, 102, 101, 105, 106, 103],
    "amount": [250, 180, 320, 450, 200, 150]
})

customers
```

```python
orders
```

```python
# Buyurtmalar jadvaliga mijozlarning ismi va shahrini qo‘shib, bitta umumiy jadval hosil qiling
orders_customers = pd.merge(orders, customers, on='customer_id')
orders_customers
```

```python
products = pd.DataFrame({
    "product_id": [1, 2, 3, 4, 5],
    "product": ["Laptop", "Telefon", "Tablet", "Monitor", "Keyboard"],
    "category": ["Computer", "Mobile", "Mobile", "Computer", "Accessory"],
    "price": [800, 500, 300, 250, 50]
})
sales = pd.DataFrame({
    "sale_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    "product_id": [1, 2, 1, 3, 5, 2, 6],
    "quantity": [2, 3, 1, 4, 10, 2, 5]
})
products
```

```python
sales
```

```python
sales_products = pd.merge(sales, products, on='product_id')
sales_products
```

```python
sales_products['quantity'] * sales_products['price']
```
