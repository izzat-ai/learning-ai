```python
import numpy as np
import pandas as pd
```

```python
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],
    "name": ["Ali", "Vali", "Hasan", "Sardor", "Madina"],
    "city": ["Toshkent", "Samarqand", "Buxoro", "Toshkent", "Andijon"]
})

orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6],
    "customer_id": [101, 102, 101, 105, 106, 103],
    "amount": [250, 180, 320, 450, 200, 150]
})

customers
```

```python
orders
```

```python
# Buyurtmalar jadvaliga mijozlarning ismi va shahrini qo‘shib, bitta umumiy jadval hosil qiling
orders_customers = pd.merge(orders, customers, on='customer_id')
orders_customers
```

**2-topshiriq**

```python
products = pd.DataFrame({
    "product_id": [1, 2, 3, 4, 5],
    "product": ["Laptop", "Telefon", "Tablet", "Monitor", "Keyboard"],
    "category": ["Computer", "Mobile", "Mobile", "Computer", "Accessory"],
    "price": [800, 500, 300, 250, 50]
})
sales = pd.DataFrame({
    "sale_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    "product_id": [1, 2, 1, 3, 5, 2, 6],
    "quantity": [2, 3, 1, 4, 10, 2, 5]
})
products
```

```python
sales
```

```python
# Savdo ma’lumotlarini mahsulot ma’lumotlari bilan birlashtiring
sales_products = pd.merge(sales, products, on='product_id')
sales_products
```

```python
# Har bir savdo uchun umumiy savdo summasini hisoblang
sales_products['quantity'] * sales_products['price']
```

**Uchinchi topshiriq**

```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105],
    "name": ["Ali", "Vali", "Hasan", "Sardor", "Madina"],
    "group": ["A", "A", "B", "B", "C"]
})
courses = pd.DataFrame({
    "course_id": [1, 2, 3, 4],
    "course": ["Python", "Statistics", "SQL", "Machine Learning"],
    "teacher": ["Anvar", "Dilshod", "Anvar", "Kamola"]
})

enrollments = pd.DataFrame({
    "student_id": [101, 101, 102, 103, 103, 104, 106],
    "course_id": [1, 2, 1, 2, 3, 4, 1],
    "score": [92, 85, 88, 79, 91, 95, 70]
})

students
```

```python
courses
```

```python
enrollments
```

```python
# students va enrollments jadvallarini birlashtirish
st_enroll = pd.merge(students, enrollments, on='student_id')
st_enroll
```

```python
# yuqorida birlashtirilgan jadvalga uchinchi DF ni qo'shish
three_df = pd.merge(st_enroll, courses, on='course_id')
three_df
```

```python
# talabaning ismi, guruhi, kurs nomi , o'qituvchisi , balini olish
three_df[['name', 'group', 'course', 'teacher', 'score']]
```

```python
# Hosil bo‘lgan jadvaldan Ali qaysi kurslarda o‘qiyotgani va har bir kursdan olgan balini ko‘rish mumkin bo‘lsin
ali = three_df[three_df['name'] == 'Ali']
ali[['course', 'score']]
```

```python
# course_id = 4 bo‘lgan kursga yozilgan talabaning ismi va bahosini toping
three_df[three_df['course_id'] == 4][['name', 'score']]
```

**To'rtinchi topshiriq**

```python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4, 5],
    "customer_name": ["Aziza", "Bekzod", "Dilnoza", "Jasur", "Malika"],
    "region": ["Toshkent", "Samarqand", "Toshkent", "Buxoro", "Andijon"]
})

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006],
    "customer_id": [1, 2, 1, 5, 7, 3],
    "product_id": [101, 102, 103, 101, 104, 102],
    "quantity": [2, 1, 3, 2, 5, 1]
})

products = pd.DataFrame({
    "product_id": [101, 102, 103, 104, 105],
    "product_name": ["Laptop", "Telefon", "Tablet", "Monitor", "Keyboard"],
    "category": ["Computer", "Mobile", "Mobile", "Computer", "Accessory"],
    "price": [800, 500, 300, 250, 50]
})

customers
```

```python
orders
```

```python
products
```

```python
# customers va orders jadvallarini birlashtirish
custom_orders = pd.merge(customers, orders, on='customer_id')
custom_orders
```

```python
# yuqoridagi jadvalga products jadvalini qo'shish
main_df = pd.merge(custom_orders, products, on='product_id')
df1 = main_df[['order_id', 'customer_name', 'region', 'product_name', 'category', 'quantity', 'price']]
df1
```

```python
# har bir buyurtma uchun umumiy buyurtma summasini hisoblang
main_df['quantity']*main_df['price']
```

```python
# Hech qanday buyurtma qilmagan mijozlar ham yakuniy jadvalda ko‘rinsin
pd.merge(customers, orders, on='customer_id', how='left')
```

```python
# Hech qachon sotilmagan mahsulotlar ham yakuniy natijada ko‘rinsin
pd.merge(custom_orders, products, on='product_id', how='right')
```

```python
# customer_id = 7 bo‘lgan buyurtmani tekshiring
pd.merge(customers, orders, on='customer_id', how='right')
```

```python
# product_id = 105 bo‘lgan mahsulotni tekshiring
pd.merge(custom_orders, products, on='product_id', how='right')
```

```python
# Mijoz + buyurtma + mahsulot — hammasi mavjud
main_df[['customer_name', 'order_id', 'product_name']]
```

```python
# Mijoz mavjud, lekin buyurtmasi yo‘q
pd.merge(customers, orders, on='customer_id', how='left')
```

```python
# Buyurtma mavjud, lekin mijoz ma'lumoti topilmagan
pd.merge(customers, orders, on='customer_id', how='right')
```

**Beshinchi topshiriq**

```python
students = pd.DataFrame({
    "name": ["Ali", "Vali", "Hasan", "Husan"],
    "group": ["A", "B", "A", "C"]
}, index=[101, 102, 103, 104])

scores = pd.DataFrame({
    "score": [92, 85, 78],
    "exam": ["Midterm", "Midterm", "Midterm"]
}, index=[101, 102, 103])
students
```

```python
scores
```

```python
# Talabalar va ularning imtihon natijalarini indekslar orqali birlashtiring
# Talabalar ro‘yxatidagi barcha talabalar saqlanishi kerak
pd.merge(students, scores, left_index=True, right_index=True, how='left')
```

####**Join**

```python
students = pd.DataFrame({
    "name": ["Ali", "Vali", "Hasan", "Husan"],
    "group": ["A", "B", "A", "C"]
}, index=[101, 102, 103, 104])

scores = pd.DataFrame({
    "score": [90, 85, 78],
    "attendance": [95, 88, 92]
}, index=[101, 102, 103])
students
```

```python
scores
```

```python
# sutdents asosiy bo'lsin va scores bilan indeks bo'yicha bog'lansin
students.join(scores, how='left')
```

```python
employees = pd.DataFrame({
    "name": ["Ali", "Vali", "Hasan", "Husan"],
    "department": ["IT", "HR", "IT", "Finance"]
}, index=[1, 2, 3, 4])

salary = pd.DataFrame({
    "salary": [800, 1000, 1200, 1500],
    "bonus": [100, 150, 200, 250]
}, index=[2, 3, 5, 6])
employees
```

```python
salary
```

```python
# xodimlar va maoshlarni birlashtiring , faqat ikkala DataFrame’da ham mavjud bo‘lgan xodimlar qolishi kerak
employees.join(salary, how='inner')
```
