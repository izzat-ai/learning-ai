**Concat**

```python
import numpy as np
import pandas as pd
```

**birinchi topshiriq**

```python
df1 = pd.DataFrame({
    "name": ["Ali", "Vali", "Hasan"],
    "age": [20, 21, 22],
    "group": ["A", "B", "A"]
})

df2 = pd.DataFrame({
    "name": ["Husan", "Sardor"],
    "age": [23, 20],
    "group": ["C", "B"]
})
df1
```

```python
df2
```

```python
# df1 va df2dagi talabalarni bitta umumiy DataFramega birlashtiring
# Talabalar ustunlari bir xil. Natijada barcha 5 ta talaba bo‘lishi kerak
pd.concat([df1, df2]).sort_index(ascending=True)
```

**ikkinchi topshiriq**

```python
students = pd.DataFrame({
    "name": ["Ali", "Vali", "Hasan"],
    "group": ["A", "B", "A"]
})

scores = pd.DataFrame({
    "math": [90, 85, 78],
    "english": [88, 92, 80]
})
students
```

```python
scores
```

```python
# students va scores ma'lumotlarini bitta DataFramega birlashtiring
pd.concat([students, scores], axis=1)
```

**uchinchi topshiriq**

```python
january = pd.DataFrame({
    "product": ["Laptop", "Phone", "Tablet"],
    "quantity": [5, 10, 7],
    "revenue": [5000, 8000, 3500]
})

february = pd.DataFrame({
    "product": ["Laptop", "Phone", "Tablet"],
    "quantity": [8, 12, 9],
    "revenue": [8000, 9600, 4500]
})

march = pd.DataFrame({
    "product": ["Laptop", "Phone", "Tablet"],
    "quantity": [6, 15, 11],
    "revenue": [6000, 12000, 5500]
})
january
```

```python
february
```

```python
march
```

```python
# Uch oyga tegishli sotuv ma'lumotlarini bitta umumiy DataFramega birlashtiring
pd.concat([january, february, march])
```

**to'rtinchi topshiriq**

```python
group_a = pd.DataFrame({
    "name": ["Ali", "Vali", "Hasan"],
    "age": [20, 21, 22],
    "score": [85, 90, 78]
})

group_b = pd.DataFrame({
    "name": ["Husan", "Sardor", "Jasur"],
    "age": [23, 20, 22],
    "score": [88, 76, 95]
})
group_a
```

```python
group_b
```

```python
# Ikki guruhdagi talabalarni bitta umumiy DataFramega birlashtiring
pd.concat([group_a, group_b], axis=0).reset_index(drop=True)
```
